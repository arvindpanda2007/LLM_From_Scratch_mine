BPE tokenizer or byte-pair encoding tokenizer is based upon a sophisticated tokenization scheme which was used to train llms such as ChatGPT

It utilizes the <| endoftext |> token to join two different segments of text

the BPE tokenizer used to train GPT-2,3 and the original model in ChatGPT had a vocabulary size of 50,257 with <| endoftext |> being assigned the largest token id of 50,256

It utilizes an algorithm which breaks down the unknown words not in its vocabulary into a sequence of smaller sub-words or even individual characters to encode it.

Example:


```python
integers = tokenizer.encode("Kabjngvf :;wfwg")
print(integers)
strings = tokenizer.decode(integers)
print(strings)
```

As we see the "unknown" string was broken down by the tokenizer into a sequence of smaller segments which were effectively decoded too later on

```python
import torch
```

```python
class GPTDataset(Dataset):
    def __init__(self, text, tokenizer, context_size, stride):
        self.input_ids = []
        self.target_ids = []
        tokenized_text = tokenizer.encode(text)
        for i in range(0, len(tokenized_text) - context_size, stride):
            input_id = tokenized_text[i:i+context_size]
            target_id = tokenized_text[i+1:i+context_size+1]
            self.input_ids.append(torch.tensor(input_id))
            self.target_ids.append(torch.tensor(target_id))

    def __len__(self):
        return len(self.input_ids)

    def __getitem__(self, sample_id):
        return self.input_ids[sample_id], self.target_ids[sample_id]
```

The next step after tokenization of raw text sample is to convert them into input-target pairs for training of an llm

for this we will implement two classes
1. Dataset class that inherits __init__ , __len__ , __getitem__ methods from Dataset class in Pytorch library
    a. __init__ method involves taking the text encodings list and initializing input_id list and a target_id list which would later be filled with tensors of inputs and targets respectively
    Note: The tensor arrays input_id and target_id is initially just two lists with tensor elements but due to torch internal mechanism of torch.stack() the entire thing is converted to a tensor

    b. __len__ returns the total number of elements in input_id or target_id or alternatively the total number of input-target pairs

    c. __getitem__ method takes the sample id that is the element number of input_id and returns the corresponding input - target pair

Now after generating the pairs we need to convert this data into batches of input(x) and target(y) stacked together
**Why?**
Because if each pair were dealt with individually then it would really be a lot of waste of computational power since GPUs are built for parallelism
For this we will use the DataLoader class
2. Dataloader class uses the Dataset class previously created and creates batches using Dataloader class in pytorch library


##IMPORTANT PARAMETERS

1. Batch Size: How many samples per batch

2. Context Window: How many tokens can the llm see before predicting next word

3. Shuffle: Shuffling in Dataloader shuffles the order of input and target in their respective batches. It is preferred since it prevents memorization of order

4. Drop_Last: Suppose there is a case where distribution of samples in each input- target pair is same but in the last pair there are very less samples left so distribution becomes lesser then drop last removes it

5. num_workers: num_workers is the number of separate CPU processes PyTorch uses to prepare data in parallel.

6. Stride: stride controls how far the sliding window moves when creating training sequences.

##How these Parameters Affect the Process

*   Larger Batch size improves GPU utilization and faster training but also increases memory usage.
*   Larger Context size improves the models reasoning capability and understanding longer contexts but also increases attention cost.
*   Smaller Stride means more training samples but also means more overlap.
*   Shuffling leads to better convergence and better prevents overfitting of patterns.
*   drop_last = true means there are fixed tensor shapes and stable GPU performance
*   more workers means faster batch preparation and better GPU utilization but increases overhead RAM usage.





##CREATING TOKEN EMBEDDINGS

The last step in preparing input text for LLM training is to convert token IDs to embedding vectors.

The first step is to initialize embedding weights with random values.
Then the model optimizes the vectors based on its discovery of percieved relationships  using backpropagation.

**Now what is backpropagation?**
During the training process the model makes predictions for the next word and then checks the actual result. If they do not match then the model goes back adjusts its internal weights and repeats the procedures. This is called backpropagation mechanism.

```python
from torch.utils.data import Dataset, DataLoader
```

```python
class dataset(Dataset):
  def __init__(self, text, tokenizer, context_size, stride):
    self.input_ids = []
    self.target_ids = []
    tokenized_text = tokenizer.encode(text)
    for i in range(0, len(tokenized_text) - context_size, stride):
      input_id = tokenized_text[i:i+context_size]
      target_id = tokenized_text[i+1:i+context_size+1]
      self.input_ids.append(torch.tensor(input_id))
      self.target_ids.append(torch.tensor(target_id))
    return None

  def __len__(self):
    return len(self.input_ids)

  def __getitem__(self, sample_id):
    return self.input_ids[sample_id], self.target_ids[sample_id]
```

```python
import torch
inputs = torch.tensor(
  [[0.43, 0.15, 0.89], # Your    (x^1)
   [0.55, 0.87, 0.66], # journey  (x^2)
   [0.57, 0.85, 0.64], # starts  (x^3)
   [0.22, 0.58, 0.33], # with   (x^4)
   [0.77, 0.25, 0.10], # one    (x^5)
   [0.05, 0.80, 0.55]] # step   (x^6)
)
```

```python
query = inputs[1]
attn_scores_2 = torch.empty(inputs.shape[0])
for i, x_i in enumerate(inputs):
    attn_scores_2[i] = torch.dot(x_i, query)
print(attn_scores_2)
```

```python
attn_weights_2_tmp = attn_scores_2 / attn_scores_2.sum()
print("Attention weights:", attn_weights_2_tmp)
print("Sum:", attn_weights_2_tmp.sum())
```

```python
def softmax_naive(x):
    return torch.exp(x) / torch.exp(x).sum(dim=0)
attn_weights_2_naive = softmax_naive(attn_scores_2)
print("Attention weights:", attn_weights_2_naive)
print("Sum:", attn_weights_2_naive.sum())
```

```python
#the softmax function is a better way to normalize attention weights since it gives higher values to important tokens and doesnt just distribute it to irrelevant info.
```

```python
'''Note that this naive softmax implementation (softmax_naive) may encounter
numerical instability problems, such as overflow and underflow, when dealing with
large or small input values. Therefore, in practice, it’s advisable to use the PyTorch
implementation of softmax, which has been extensively optimized for performance:'''
```

```python
attn_weights_2 = torch.softmax(attn_scores_2, dim=0)
print("Attention weights:", attn_weights_2)
print("Sum:", attn_weights_2.sum())
```

```python
#This is the final representation of the context token
query = inputs[1]
context_vec_2 = torch.zeros(query.shape)
for i,x_i in enumerate(inputs):
    context_vec_2 += attn_weights_2[i]*x_i
print(context_vec_2)
```

```python
attn_scores = torch.empty(6, 6)
for i, x_i in enumerate(inputs):
    for j, x_j in enumerate(inputs):
        attn_scores[i, j] = torch.dot(x_i, x_j)
print(attn_scores)
```
