BPE tokenizer or byte-pair encoding tokenizer is based upon a sophisticated tokenization scheme which was used to train llms such as ChatGPT 

It utilizes the <| endoftext |> token to join two different segments of text 

the BPE tokenizer used to train GPT-2,3 and the original model in ChatGPT had a vocabulary size of 50,257 with <| endoftext |> being assigned the largest token id of 50,256

It utilizes an algorithm which breaks down the unknown words not in its vocabulary into a sequence of smaller sub-words or even individual characters to encode it.

Example:


```python
integers = tokenizer.encode("Kabjngvf :;wfwg")
print(integers)
strings = tokenizer.decode(integers)
print(strings)
```

As we see the "unknown" string was broken down by the tokenizer into a sequence of smaller segments which were effectively decoded too later on

```python
import tiktoken
tokenizer = tiktoken.get_encoding("gpt2")
with open("the-verdict.txt", "r", encoding="utf-8") as f:
    raw_text = f.read()
enc_text = tokenizer.encode(raw_text)
print(len(enc_text))
```

```python
enc_sample = enc_text[50:]

context_size = 5      
x = enc_sample[:context_size]
y = enc_sample[1:context_size+1]
print(f"x: {x}")
print(f"y:      {y}")
```

```python
for i in range(1, context_size+1):
    context = enc_sample[:i]
    desired = enc_sample[i]
    print(context, "---->", desired)
```

```python
for i in range(1, context_size+1):
    context = enc_sample[:i]
    desired = enc_sample[i]
    print(tokenizer.decode(context), "---->", tokenizer.decode([desired]))
```

```python
from torch.utils.data import Dataset, DataLoader
```

```python
import tiktoken
with open("the-verdict.txt", "r", encoding="utf-8") as f:
    raw_text = f.read()
tokenizer = tiktoken.get_encoding("gpt2")
```

```python
import torch
```

```python
class GPTDataset(Dataset):
    def __init__(self, text, tokenizer, context_size, stride):
        self.input_ids = []
        self.target_ids = []
        tokenized_text = tokenizer.encode(text)
        for i in range(0, len(tokenized_text) - context_size, stride):
            input_id = tokenized_text[i:i+context_size]
            target_id = tokenized_text[i+1:i+context_size+1]
            self.input_ids.append(input_id)
            self.target_ids.append(target_id)

    def __len__(self):
        return len(self.input_ids)

    def __getitem__(self, sample_id):
        return self.input_ids[sample_id], self.target_ids[sample_id]
```

```python
dataset = GPTDataset(
    raw_text, tokenizer,
    context_size=4,
    stride=1
)
```

```python
def data_loader(text, batch_size, context_window = 256, stride = 64, shuffle = False,num_workers = 3):
    tokenizer = tiktoken.get_encoding("gpt2")
    dataset = GPTDataset(
        text, tokenizer,
        context_size=context_window,
        stride=stride
    )
    return DataLoader(dataset, batch_size=batch_size, shuffle=shuffle, num_workers=num_workers)
    
```

```python
dataloader = data_loader(raw_text, batch_size=4, context_window = 256, stride = 64, shuffle=True, num_workers=14)
```
